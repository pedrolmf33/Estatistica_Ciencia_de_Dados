### Importação e Conhecimento dos Dados

```python
# Bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carregando CSV
df = pd.read_csv('SeoulBikeData.csv', encoding='latin1')
```

```python
print(df.shape)
print('-------------------------------')
print(df.dtypes)
print('-------------------------------')
df.info()
print('-------------------------------')
df.head()
```

```python
# Padronizando os nomes das colunas
df.columns = (df.columns.str.lower().str.replace(' ', '_').str.strip())
df.head()
```

```python
# Converter a coluna date para o tipo datatime64
df['date'] = pd.to_datetime(df['date'], dayfirst=True)

# Criar coluna 'timestamp' combinando data + hora
df["timestamp"] = df["date"] + pd.to_timedelta(df["hour"], unit="h")

# Conferir os tipos atualizados
print(df[["date", "hour", "timestamp"]].dtypes)
print(df[["date", "hour", "timestamp"]].head())
```

```python
print("Dimensões:", df.shape)
print("Colunas:", df.columns.tolist())
```

```python
# Últimas linhas
df.tail()
```

```python
# Verificando quantidade de valores nulos e duplicados
print(f'Quantidade de valores nulos: \n', df.isnull().sum())
print('---------------')
print(f'Quantidade de valores duplicados:', df.duplicated().sum())
```

```python
# Estatística descritiva das variáveis numéricas
df.describe()
```

```python
# Estatística das variáveis categóricas
df.describe(include='object')
```

```python
# Verificando os valores únicos das colunas categóricas

for col in ['seasons', 'holiday', 'functioning_day']:
    print(f'\n Valores únicos em {col}:', df[col].unique())
    print(df[col].value_counts())
```

```python
# Checando a relação entre functioning_day e rented_bike_count

print('Resumo de Rented Bike Count por Functioning Day:\n')
print(df.groupby('functioning_day')['rented_bike_count'].agg(['count', 'min', 'max', 'median', 'mean']))
```

```python
df[df["functioning_day"] == "No"][
    ["date", "hour", "functioning_day", "rented_bike_count"]
].head(10)
```

```python
# Filtrar registro sem funcionamento
sem_func = df[df['functioning_day'] == 'No']

# Vendo quais datas estão com pelo menos um registro sem funcionamento
data_sem_func = sem_func['date'].unique()
print(data_sem_func)
```

```python
# Contando quantos registro sem funcionamento teve em cada data

resumo_datas = sem_func.groupby('date').agg(
    horas_sem_func = ('hour', 'count'),
    primeira_hora = ('hour', 'min'),
    ultima_hora = ('hour', 'max'),
    total_alugueis = ('rented_bike_count', 'sum'),
)
print(resumo_datas)
```

```python
# Comparando quantidade de bikes alugadas com e sem chuva e neve

sem_chuva = df[(df['rainfall(mm)'] == 0) & (df["functioning_day"] == "Yes")]
sem_neve = df[(df['snowfall_(cm)'] == 0) & (df["functioning_day"] == "Yes")]
com_chuva = df[(df['rainfall(mm)'] != 0) & (df["functioning_day"] == "Yes")]
com_neve = df[(df['snowfall_(cm)'] != 0) & (df["functioning_day"] == "Yes")]

alugadas_sem_chuva = sem_chuva.agg(
    horas=('rented_bike_count', 'count'),
    media=('rented_bike_count', 'mean'),
    mediana=('rented_bike_count', 'median'),
    total=('rented_bike_count', 'sum')
)

alugadas_sem_neve = sem_neve.agg(
    horas=('rented_bike_count', 'count'),
    media=('rented_bike_count', 'mean'),
    mediana=('rented_bike_count', 'median'),
    total=('rented_bike_count', 'sum')
)

alugadas_com_chuva = com_chuva.agg(
    horas=('rented_bike_count', 'count'),
    media=('rented_bike_count', 'mean'),
    mediana=('rented_bike_count', 'median'),
    total=('rented_bike_count', 'sum')
)

alugadas_com_neve = com_neve.agg(
    horas=('rented_bike_count', 'count'),
    media=('rented_bike_count', 'mean'),
    mediana=('rented_bike_count', 'median'),
    total=('rented_bike_count', 'sum')
)

print('Não estando chovendo')
print(alugadas_sem_chuva.to_string(float_format=lambda x: f'{x:,.2f}'))

print('\nNão estando nevando')
print(alugadas_sem_neve.to_string(float_format=lambda x: f'{x:,.2f}'))

print('\nEstando chovendo')
print(alugadas_com_chuva.to_string(float_format=lambda x: f'{x:,.2f}'))

print('\nEstando nevando')
print(alugadas_com_neve.to_string(float_format=lambda x: f'{x:,.2f}'))
```

### Conclusões da Etapa 1 — Importação e Conhecimento dos Dados

* O dataset possui **8.760 registros horários** e **14 variáveis originais** (cobrindo 365 dias completos, de 01/12/2017 a 30/11/2018), sem valores ausentes (`NaN`) e sem linhas duplicadas. <p>
* **Padronização e preparação:** Os nomes das colunas foram padronizados para minúsculo com *underscore* (`_`), a coluna `date` foi convertida para o tipo `datetime64` e foi criada a coluna `timestamp` (combinando data e hora) para facilitar análises cronológicas e séries temporais nas próximas etapas. <p>
* **Comportamento de funcionamento (`functioning_day`):** Foram identificados **295 registros** (distribuídos em 13 datas específicas) em que o sistema não estava em operação (`No`), todos com **0 bicicletas alugadas**. Nos **8.465 registros** operacionais (`Yes`), a demanda variou de **2 a 3.556 bicicletas/hora** (média de 729,16 e mediana de 542,00), mostrando que os zeros estão diretamente ligados aos períodos sem funcionamento e não configuram erro de registro.<p>
* **Impacto inicial de chuva e neve (em dias funcionais):** A presença de precipitação reduz drasticamente a demanda média por hora — caindo de **765,63** para **167,26** bicicletas/hora em horas com chuva, e de **759,20** para **185,10** bicicletas/hora em horas com neve.<p>
* **Outliers:** Valores extremos observados (como picos de demanda ou chuvas intensas) representam variações reais do fenômeno e foram mantidos integralmente para investigação na Análise Exploratória (Etapa 2).
